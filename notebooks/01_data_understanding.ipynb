```python
import pandas as pd
import numpy as np

# Load datasets
leads = pd.read_csv("../data/raw/olist_marketing_qualified_leads_dataset.csv")
deals = pd.read_csv("../data/raw/olist_closed_deals_dataset.csv")

print("Marketing Qualified Leads Dataset:")
print("Shape:", leads.shape)
display(leads.head())

print("\nClosed Deals Dataset:")
print("Shape:", deals.shape)
display(deals.head())
```

```python
# Check Marketing Qualified Leads
print("MARKETING QUALIFIED LEADS")
print("\nColumn Names:")
print(leads.columns.tolist())

print("\nData Types:")
print(leads.dtypes)

print("\nMissing Values:")
print(leads.isnull().sum())

print("\nDuplicate Rows:", leads.duplicated().sum())


# Check Closed Deals
print("\n\nCLOSED DEALS")
print("\nColumn Names:")
print(deals.columns.tolist())

print("\nData Types:")
print(deals.dtypes)

print("\nMissing Values:")
print(deals.isnull().sum())

print("\nDuplicate Rows:", deals.duplicated().sum())
```

```python
# Lead Origin Distribution
print("LEAD ORIGIN DISTRIBUTION")
display(leads['origin'].value_counts(dropna=False))

# Business Segment Distribution
print("\nBUSINESS SEGMENT DISTRIBUTION")
display(deals['business_segment'].value_counts(dropna=False))

# Lead Type Distribution
print("\nLEAD TYPE DISTRIBUTION")
display(deals['lead_type'].value_counts(dropna=False))

# Numerical Columns Summary
print("\nNUMERICAL DATA SUMMARY")
display(deals.describe())

# Date Range
print("\nFIRST CONTACT DATE RANGE")
print(leads['first_contact_date'].min(), "to", leads['first_contact_date'].max())

print("\nWON DATE RANGE")
print(deals['won_date'].min(), "to", deals['won_date'].max())
```
