```python
import pandas as pd
import numpy as np

# Load original datasets
leads = pd.read_csv("data/raw/olist_marketing_qualified_leads_dataset.csv")
deals = pd.read_csv("data/raw/olist_closed_deals_dataset.csv")

print("Leads Shape:", leads.shape)
print("Deals Shape:", deals.shape)
```

```python
# Check and remove duplicates
leads = leads.drop_duplicates()
deals = deals.drop_duplicates()

# Final shape
print("Final Leads Shape:", leads.shape)
print("Final Deals Shape:", deals.shape)

# Save cleaned datasets
leads.to_csv("data/processed/cleaned_marketing_leads.csv", index=False)
deals.to_csv("data/processed/cleaned_closed_deals.csv", index=False)

print("\nCleaned datasets exported successfully!")
```
